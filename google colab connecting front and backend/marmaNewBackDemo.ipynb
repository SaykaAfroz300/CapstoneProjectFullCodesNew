### Re-establishing Ngrok Tunnel

If you are experiencing connectivity issues (e.g., 'no website'), the ngrok tunnel might have become inactive. The cell below will close any existing tunnels and create a new one, providing an updated public URL. Please **ensure you have stopped any running FastAPI server cells** (e.g., by clicking the square stop button next to `await server.serve()`) before running this cell.

```python
import os
from pyngrok import ngrok

# Kill any existing ngrok tunnels to ensure a clean slate
ngrok.kill()

# Open a new tunnel on port 8000 (where the FastAPI app is running)
public_url_new = ngrok.connect(8000).public_url

print("="*60)
print(f"🌟 NEW API IS LIVE AT: {public_url_new}")
print("Update the 'API_URL' in your HTML file to:")
print(f"const API_URL = '{public_url_new}/translate';")
print("=")
print("If you continue to experience connection issues, please verify your local network firewall settings.")
print("=")
print("On iOS, if you see a security warning or a prompt to 'visit site', click through it. This is usually a browser security feature.")
print("="*60)
```
